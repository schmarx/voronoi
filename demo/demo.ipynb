```python
import matplotlib.pyplot as plt
import numpy as np

import subprocess
```

```python
def get_point(string):
    return [float(coord) for coord in string[1:-1].split(", ") ]

def get_points(string):
    points = string.split("\t")
    if len(points) < 2:
        return []
    
    point1 = get_point(points[0])
    point2 = get_point(points[1])

    return [point1, point2]

subprocess.run(["make", "-C", "../"])


file = open("./output/edges.txt", "r")
contents = file.read()

point_data, edge_data = contents.split("\n", 1)
edges = []
points = []

for point in point_data.split("\t"):
    pt = get_point(point)
    points.append(pt)

for edge in edge_data.split("\n"):
    point = get_points(edge)
    if len(point) < 2:
        continue

    # if (point1[0] > 20 or point1[1] > 20):
    #     continue
    # if (point2[0] > 20 or point2[1] > 20):
    #     continue
    # if (point1[0] < -20 or point1[1] < -20):
    #     continue
    # if (point2[0] < -20 or point2[1] < -20):
    #     continue
    # if (point[0][1] < 0 or point[1][1] < 0):
    #     continue

    edges.append((point[0], point[1]))

fig, ax = plt.subplots(1, figsize=(10, 10))

# plt.Polygon()

for edge in edges:
    ax.plot((edge[0][0], edge[1][0]), (edge[0][1], edge[1][1]), ".-", color="blue")

for point in points:
    ax.plot(point[0], point[1], ".", color="red")
```
